```python
from business_line import BusinessLine
import pandas as pd

try:
    bl = BusinessLine('GKK')

    engine = bl.create_engine()
    conn = engine.connect()

    sql = """
        SELECT
	        CUSTOMERID
	        , FIRSTNAME
        FROM 
            CORE_PROD.DATAMART_GKK.CUSTOMER_DIM_STD_GKK c
        LIMIT 10
        """

    df = pd.read_sql(sql, conn)

finally:
    conn.close()
    engine.dispose()

df
```
